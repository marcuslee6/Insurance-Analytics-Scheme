```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.linear_model import Perceptron
from sklearn.svm import SVC
from sklearn.metrics import classification_report

# 1. 讀取並處理數據
df = pd.read_csv('insurance.csv')

# 定義目標：將收費高於中位數嘅定為 1 (High Risk)，否則為 0 (Low Risk)
df['risk_level'] = (df['charges'] > df['charges'].median()).astype(int)

# 選擇特徵並轉換類別變數
df['smoker'] = df['smoker'].map({'yes': 1, 'no': 0})
X = df[['age', 'bmi', 'children', 'smoker']]
y = df['risk_level']

# 2. 數據標準化 (對 Perceptron 和 SVM 至關重要)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 分割訓練集同測試集
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.25, random_state=36)

# 3. LDA (降維與分類)
lda = LDA(n_components=1)
X_train_lda = lda.fit_transform(X_train, y_train)
X_test_lda = lda.transform(X_test)

# 4. Perceptron (單層感知器 - 模擬簡單神經網絡邏輯)
ppn = Perceptron(max_iter=1000, eta0=0.01, random_state=36)
ppn.fit(X_train, y_train)
y_pred_ppn = ppn.predict(X_test)

# 5. SVM (支持向量機 - 最大化邊界)
svm = SVC(kernel='linear', C=5.0, random_state=36)
svm.fit(X_train, y_train)
y_pred_svm = svm.predict(X_test)

# 6. 顯示結果
print("--- Perceptron Result ---")
print(classification_report(y_test, y_pred_ppn))
print("\n--- SVM Result ---")
print(classification_report(y_test, y_pred_svm))
```
